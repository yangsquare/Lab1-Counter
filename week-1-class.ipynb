Here is a little sample of plotting timings for a function in Python.

````python
# this first line is just used to make the plots appear nicely in the notebook
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import time

def fake_function(n):
    time.sleep(0.001*n*n)

def timeit(n):
    start = time.time()
    fake_function(n)
    end = time.time()
    return end-start

N = [0, 10, 20, 30, 40, 50]
T = [timeit(n) for n in N]
plt.plot(N, T)
plt.xlabel('n')
plt.ylabel('Time (s)');
````

````python
def fib_recursive(n):
  if n<=1:
    return n
  return fib_recursive(n-1)+fib_recursive(n-2)

def fib_table(n):
  if n<=1:
    return n

  table=[0,1]
  for i in range(2, n+1):
    table.append(table[i-1]+table[i-2])
  return table[n]

def fib_last_two(n):
  a=0
  b=1
  for i in range(n):
    a,b=b,a+b
  return a

print(fib_last_two(10))
print(fib_recursive(10))
print(fib_table(10))
````

````python
import time
import matplotlib.pyplot as plt

def timer(func, n, repeats=1):
  start =time.perf_counter()

  for i in range(repeats):
    func(n)
  end=time.perf_counter()
  return (end-start)/repeats

N1=list(range(1,31))
T1=[timer(fib_recursive,n) for n in N1]

plt.figure(figsize=(8,4))
plt.plot(N1, T1)
plt.xlabel("n")
plt.ylabel("Time in seconds")
plt.title("recursive fib")
plt.grid()
plt.show

N2=list(range(1,101))
T3=[timer(fib_last_two,n,1000) for n in N2]
T2=[timer(fib_table,n,1000) for n in N2]

plt.figure(figsize=(8,4))
plt.plot(N2, T2, label="table")
plt.plot(N2, T3, label="last two")
plt.xlabel("n")
plt.ylabel("time in seconds")
plt.title("fib comparison")
plt.legend()
plt.grid()
plt.show()

````

````python
from decimal import Decimal, localcontext

def fib_analytic(n):
  with localcontext() as ctx:
    ctx.prec=100
    root5 =Decimal(5).sqrt()
    phi=(1+root5)/2
    psi=(1-root5)/2
    answer=(((phi**n)-(psi**n))/root5)
    return int(answer.to_integral_value())

print(fib_analytic(10))


T4=[timer(fib_analytic, n, 20) for n in N2]
plt.figure(figsize=(8,4))
plt.plot(N2, T4, label="analytic")

plt.xlabel("n")
plt.ylabel("time in seconds")
plt.title("analytic")
plt.legend()
plt.grid()
plt.show()
````

## Implementation notes on compiled languages

For the other ones, if you're using a compiled language, they might run so fast that in order to time them you should slow down the addition by adding a 1 microsecond wait. I've included a C++ ``busy_sleep`` function below to do that.

```c++
#include <chrono>
#include <iostream>
#include <fstream>

using namespace std;

// this function busy sleeps for n microseconds
void busy_sleep(int n)
{
    auto start = chrono::steady_clock::now();
    while(chrono::duration_cast<chrono::microseconds>(chrono::steady_clock::now() - start).count()<n) {};
}
```
